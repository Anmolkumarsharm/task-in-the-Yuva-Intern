```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x ="Fertilizer_Used(tons)",y ='Yield(tons)',hue='Crop_Type')
plt.title('Fertilizer used vs Crop Yield')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(),annot = True,cmap='coolwarm',fmt='.2f')
plt.title('Features Correlations heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x= 'Crop_Type', order=df['Crop_Type'].value_counts().index)
plt.title('Distribution of Crop Types')
plt.xticks(rotation=45)
plt.show()
```
